```python
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

dummy = evaluate(DummyClassifier(strategy='most_frequent'), "Dummy")

logit = evaluate(make_pipeline(SimpleImputer(strategy='median'), StandardScaler(),
                 LogisticRegression(max_iter=1000, class_weight='balanced')), "Logistic Regression")
```

```python
import pandas as pd
pd.DataFrame([dummy, logit], index=["Dummy", "Logistic Regression"]).to_csv("results_logistic_regression.csv")
```
